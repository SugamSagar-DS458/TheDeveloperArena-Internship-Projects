```python
import pandas as pd

# Load the sales data into a DataFrame
df = pd.read_csv(r'C:\Users\sumit\DATA_SCIENCE\THE DEVELOPER ARENA INTERNSHIP\Week 3\sales_data.csv')
```

```python
# Display the first 5 rows of the DataFrame
display(df.head())
```

```python
# Display the shape (number of rows and columns) of the DataFrame
print(f"DataFrame shape: {df.shape}")
```

```python
# Display the column names of the DataFrame
print("\nDataFrame columns:")
for col in df.columns:
    print(f"- {col}")
```

```python
# Display a concise summary of the DataFrame, including data types and non-null values
print("\nDataFrame info:")
df.info()
```

```python
# Check for duplicate rows
duplicate_rows = df.duplicated().sum()
print(f"Number of duplicate rows: {duplicate_rows}")

# If duplicates exist, remove them
if duplicate_rows > 0:
    df.drop_duplicates(inplace=True)
    print("Duplicate rows removed.")
    print(f"New DataFrame shape after removing duplicates: {df.shape}")
```

```python
# Calculate the total revenue
total_revenue = df['Total_Sales'].sum()
print(f"Total Revenue: ${total_revenue:,.2f}")
```

```python
# Find the best-selling product by total sales
best_selling_product = df.groupby('Product')['Total_Sales'].sum().idxmax()
max_product_sales = df.groupby('Product')['Total_Sales'].sum().max()
print(f"Best Selling Product: {best_selling_product} (Total Sales: ${max_product_sales:,.2f})")
```

```python
# Find the best-performing region by total sales
best_performing_region = df.groupby('Region')['Total_Sales'].sum().idxmax()
max_region_sales = df.groupby('Region')['Total_Sales'].sum().max()
print(f"Best Performing Region: {best_performing_region} (Total Sales: ${max_region_sales:,.2f})")
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate total sales per product
product_sales = df.groupby('Product')['Total_Sales'].sum().reset_index()

# Sort for better visualization
product_sales = product_sales.sort_values(by='Total_Sales', ascending=False)

plt.figure(figsize=(5, 4))
sns.barplot(x='Product', y='Total_Sales', data=product_sales, palette='viridis')
plt.title('Total Sales by Product')
plt.xlabel('Product')
plt.ylabel('Total Sales ($)')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate total sales per region
region_sales = df.groupby('Region')['Total_Sales'].sum().reset_index()

# Sort for better visualization
region_sales = region_sales.sort_values(by='Total_Sales', ascending=False)

plt.figure(figsize=(5, 4))
sns.barplot(x='Region', y='Total_Sales', data=region_sales, palette='magma')
plt.title('Total Sales by Region')
plt.xlabel('Region')
plt.ylabel('Total Sales ($)')
plt.tight_layout()
plt.show()
```
