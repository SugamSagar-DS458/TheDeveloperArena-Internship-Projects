```python
def calculate_grade(marks):
    # Grading Logic using if-elif-else statements 
    if marks >= 90: 
        grade = "A" 
        message = "Excellent! You are doing amazing work! 🌟" 
    elif marks >= 80: 
        grade = "B" 
        message = "Very Good! Keep it up! 👍" 
    elif marks >= 70: 
        grade = "C" 
        message = "Good job! With a little more effort, you can reach the top! 💪" 
    elif marks >= 60: 
        grade = "D" 
        message = "Passed. Let's work harder next time to boost this score! 📈" 
    else: 
        grade = "F" 
        message = "Don't give up! Reach out for help and try again. You can do it! ❤️" 
    return grade, message 

def main(): 
    print("--- Student Grade Calculator ---") 
    name = input("Enter student name: ").strip() 
    
    # Use while loop to handle invalid inputs and add input validation 
    while True: 
        try: 
            marks_input = input("Enter marks (0-100): ") 
            marks = float(marks_input) 
            # Check marks range 
            if 0 <= marks <= 100: 
                break # Valid input, exit the loop 
            else: 
                print("Invalid input. Marks must be between 0 and 100.") 
        except ValueError: 
            print("Invalid input. Please enter a valid number.") 
            
    # Calculate grade using the function 
    grade, message = calculate_grade(marks) 
    
    # Print the formatted sample output 
    print(f"\n📊 RESULT FOR {name.upper()}:") 
    print(f"Marks: {marks:g}/100") 
    print(f"Grade: {grade}") 
    print(f"Message: {message}") 

# Run the program 
if __name__ == "__main__": 
    main()
```
