```python
import pandas as pd
import numpy as np
```

```python
# Load the dataset from the original CSV
df_cleaned_for_encoding = pd.read_csv(r'C:\Users\sumit\DATA_SCIENCE\THE DEVELOPER ARENA INTERNSHIP\Week 10\customer_churn.csv')

# Drop CustomerID as it's not needed for modeling
df_cleaned_for_encoding = df_cleaned_for_encoding.drop('CustomerID', axis=1)

# Convert 'TotalCharges' to numeric, handling missing values by filling with 0
df_cleaned_for_encoding['TotalCharges'] = pd.to_numeric(df_cleaned_for_encoding['TotalCharges'], errors='coerce')
df_cleaned_for_encoding['TotalCharges'] = df_cleaned_for_encoding['TotalCharges'].fillna(0)

# Convert 'PaperlessBilling' to numerical (0 or 1)
df_cleaned_for_encoding['PaperlessBilling'] = df_cleaned_for_encoding['PaperlessBilling'].map({'Yes': 1, 'No': 0})

# Convert 'SeniorCitizen' to numerical (0 or 1) if it's not already
df_cleaned_for_encoding['SeniorCitizen'] = df_cleaned_for_encoding['SeniorCitizen'].astype(int)

# The 'Churn' column is already numerical (0s and 1s) from the original dataset, so no mapping is needed.
# df_cleaned_for_encoding['Churn'] = df_cleaned_for_encoding['Churn'].map({'Yes': 1, 'No': 0}) # Removed this line

print("DataFrame head after initial cleaning (before specific encoding demonstrations):")
display(df_cleaned_for_encoding.head())
print("\nData types after initial cleaning:")
display(df_cleaned_for_encoding.dtypes)
```

## Categorical Feature Encoding

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, OrdinalEncoder

# Make a copy of the original dataframe 'df' for encoding operations.
df_encoded = df_cleaned_for_encoding.copy()

# Identify categorical columns (excluding 'PaperlessBilling' and 'SeniorCitizen' if they were already handled or are numerical).
categorical_cols = df_encoded.select_dtypes(include='object').columns.tolist()

print(f"Categorical columns to be encoded: {categorical_cols}")
```

### 1. One-Hot Encoding

```python
# Apply One-Hot Encoding using pd.get_dummies
df_onehot = pd.get_dummies(df_encoded, columns=categorical_cols, drop_first=True)

print("DataFrame after One-Hot Encoding:")
display(df_onehot.head())
```

### 2. Label Encoding

```python
# Apply Label Encoding
df_label_encoded = df_encoded.copy()
label_encoder = LabelEncoder()

for col in categorical_cols:
    df_label_encoded[col] = label_encoder.fit_transform(df_label_encoded[col])

print("DataFrame after Label Encoding:")
display(df_label_encoded.head())
```

### 3. Ordinal Encoding

```python
# Apply Ordinal Encoding (assuming no specific order for now)
df_ordinal_encoded = df_encoded.copy()
ordinal_encoder = OrdinalEncoder()

df_ordinal_encoded[categorical_cols] = ordinal_encoder.fit_transform(df_ordinal_encoded[categorical_cols])

print("DataFrame after Ordinal Encoding:")
display(df_ordinal_encoded.head())
```

## Feature Engineering

```python
# Make a copy of the cleaned dataframe to add engineered features
df_engineered = df_cleaned_for_encoding.copy()

# 1. Average Charge Per Tenure (TotalCharges / Tenure)
df_engineered['AvgChargePerTenure'] = df_engineered['TotalCharges'] / df_engineered['Tenure']
# Handle division by zero for Tenure=0 (replace with 0 or MonthlyCharges)
df_engineered['AvgChargePerTenure'] = df_engineered['AvgChargePerTenure'].replace([np.inf, -np.inf], np.nan).fillna(df_engineered['MonthlyCharges'])

# 2. IsLongTermCustomer (e.g., Tenure > 24 months)
df_engineered['IsLongTermCustomer'] = (df_engineered['Tenure'] > 24).astype(int)

# 3. MonthlyChargeToTotalChargeRatio (MonthlyCharges / TotalCharges)
df_engineered['MonthlyChargeToTotalChargeRatio'] = df_engineered['MonthlyCharges'] / df_engineered['TotalCharges']
df_engineered['MonthlyChargeToTotalChargeRatio'] = df_engineered['MonthlyChargeToTotalChargeRatio'].replace([np.inf, -np.inf], np.nan).fillna(0)

# 4. TenureToMonthlyChargeRatio (Tenure / MonthlyCharges)
df_engineered['TenureToMonthlyChargeRatio'] = df_engineered['Tenure'] / df_engineered['MonthlyCharges']
df_engineered['TenureToMonthlyChargeRatio'] = df_engineered['TenureToMonthlyChargeRatio'].replace([np.inf, -np.inf], np.nan).fillna(0)

# 5. IsHighValueCustomer (based on TotalCharges)
df_engineered['IsHighValueCustomer'] = (df_engineered['TotalCharges'] > df_engineered['TotalCharges'].median()).astype(int)

print("DataFrame head with new engineered features:")
display(df_engineered.head())
```

## Complete Preprocessing Pipeline

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.model_selection import train_test_split
import pandas as pd
import numpy as np

def preprocess_data(file_path):
    # Load the dataset
    df = pd.read_csv(r"C:\Users\sumit\DATA_SCIENCE\THE DEVELOPER ARENA INTERNSHIP\Week 10\customer_churn.csv")

    # Drop CustomerID
    df = df.drop('CustomerID', axis=1)

    # Convert 'TotalCharges' to numeric, handling missing values
    df['TotalCharges'] = pd.to_numeric(df['TotalCharges'], errors='coerce')
    df['TotalCharges'] = df['TotalCharges'].fillna(0)

    # Convert 'PaperlessBilling' to numerical
    df['PaperlessBilling'] = df['PaperlessBilling'].map({'Yes': 1, 'No': 0})

    # Convert 'SeniorCitizen' to numerical
    df['SeniorCitizen'] = df['SeniorCitizen'].astype(int)

    # Convert 'Churn' to numerical (if not already)
    # Assuming Churn is already 0/1 from previous steps; otherwise, uncomment:
    # df['Churn'] = df['Churn'].map({'Yes': 1, 'No': 0})

    # Define features (X) and target (y)
    X = df.drop('Churn', axis=1)
    y = df['Churn']

    # Identify categorical and numerical columns
    categorical_features = X.select_dtypes(include=['object']).columns
    numerical_features = X.select_dtypes(include=['int64', 'float64']).columns

    # Create preprocessing pipelines for numerical and categorical features
    numerical_transformer = StandardScaler()
    categorical_transformer = OneHotEncoder(handle_unknown='ignore', drop='first') # drop='first' to avoid multicollinearity

    # Create a preprocessor using ColumnTransformer
    preprocessor = ColumnTransformer(
        transformers=[
            ('num', numerical_transformer, numerical_features),
            ('cat', categorical_transformer, categorical_features)
        ])

    # Create a full preprocessing pipeline
    pipeline = Pipeline(steps=[('preprocessor', preprocessor)])

    # Apply preprocessing to X
    X_processed = pipeline.fit_transform(X)

    # Get feature names after one-hot encoding
    ohe_feature_names = pipeline.named_steps['preprocessor'].named_transformers_['cat'].get_feature_names_out(categorical_features)
    all_feature_names = list(numerical_features) + list(ohe_feature_names)

    # Convert processed X back to a DataFrame for better interpretability
    X_processed_df = pd.DataFrame(X_processed, columns=all_feature_names)

    return X_processed_df, y

# Run the pipeline
file_path = '/content/customer_churn.csv'
X_final, y_final = preprocess_data(file_path)

print("Preprocessed Features (X_final) head:")
display(X_final.head())
print("\nTarget Variable (y_final) head:")
display(y_final.head())
```

```python
# Split the preprocessed data into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(X_final, y_final, test_size=0.2, random_state=42, stratify=y_final)

print("X_train shape:", X_train.shape)
print("X_test shape:", X_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```
